```python
import numpy as np

#Matriz hermitica
A = np.array([
    [2, 1+1j],
    [1-1j, 3]
])

print(f"A:{A}")

#Comprobar si es hermitica
def es_hermitica_con_tolerancia(matriz, tolerancia=1e-10):
    n = matriz.shape[0]

    for i in range(n):
        for j in range(n):
            #diferencia absoluta
            diferencia = abs(matriz[i,j]-np.conj(matriz[j,i]))
            if diferencia > tolerancia:
                return False
    return True

print("\n¿A es hermitica?")
print(es_hermitica_con_tolerancia(A))

#autovalores y autovectores
autovalores, autovectores = np.linalg.eigh(A)

print("\nAutovalores:")
print(autovalores)

print("\nAutovectores:")
print(autovectores)


#Producto interno
u = autovectores[:, 0]
v = autovectores[:, 1]

producto_interno = np.vdot(u, v)

print("\nProducto interno <u,v>:")
print(producto_interno)

print("\n¿Los autovectores son ortogonales?")
print(np.isclose(producto_interno, 0))
```
